# Module 4 - Multilayer Perceptron
### Demonstrate a Multilayer Perceptron with forward propagation

This module's notebook demostrates a Multilayer Perceptron (MLP) with forward propagation using NumPy. The neural network consists of the following architecture:

* Input Layer: 3 neurons
* Hidden Layer 1: 5 neurons - Sigmoid
* Hidden Layer 2: 5 neurons - Sigmoid
* Hidden Layer 3: 2 neurons - Sigmoid
* Output Later: 2 neurons - ReLU

This implementation supports both randomly generated andf manuallu specified weights and biases. The purpose of this demonstration is the verify that input data successfully propagates through all layers of the network.


```python
import numpy as np
import module4
```

### Test 1: Random Weights and Biases
Objective is to verify that the MLP successfully performs forward propagation.

```python
input_vector = module4.set_input_vector(0.5, 0.2, 0.1)

output = module4.random_mlp_weights_and_biases(input_vector)
```

### Test 2
Demonstrates the change in the output from modifying parameters

Initial setup:

```python
# Manually specified weights
w1 = np.full((3, 5), 0.1)
w2 = np.full((5, 5), 0.1)
w3 = np.full((5, 2), 0.1)
w4 = np.full((2, 2), 0.1)

# Manually specified biases
b1 = np.zeros((1, 5))
b2 = np.zeros((1, 5))
b3 = np.zeros((1, 2))
b4 = np.zeros((1, 2))

# Execute forward propagation
output = module4.set_mlp_weights_and_biases(
    input_vector,
    w1, b1,
    w2, b2,
    w3, b3,
    w4, b4
)

assert output.shape == (1, 2)
```

The following will changing the weights, keeping biases and input the same as the initial

```python
# Create a copy of the original weights
w4_modified = w4.copy()

# Modify one weight
w4_modified[0, 0] = 1.0

# Execute forward propagation
weight_output = module4.set_mlp_weights_and_biases(
    input_vector,
    w1, b1,
    w2, b2,
    w3, b3,
    w4_modified, b4
)

# Compare the results
print("Original Output:", output)
print("Modified Output:", weight_output)
print("Difference:", weight_output - output)
```

The following will change the biases, keep weights and input the same from the initial

```python
# Create a copy of the original biases
b4_modified = b4.copy()

# Modify one bias
b4_modified[0, 1] = 0.5

# Execute forward propagation
bias_output = module4.set_mlp_weights_and_biases(
    input_vector,
    w1, b1,
    w2, b2,
    w3, b3,
    w4, b4_modified
)

# Compare the results
print("Original Output:", output)
print("Modified Output:", bias_output)
print("Difference:", bias_output - output)
```

The following changes the input but keep the initial weights and biases

```python
# Define a different input vector
new_input_vector = module4.set_input_vector(1.0, 0.2, 0.1)

# Execute forward propagation with original parameters
input_output = module4.set_mlp_weights_and_biases(
    new_input_vector,
    w1, b1,
    w2, b2,
    w3, b3,
    w4, b4
)

# Compare the results
print("Original Input:", input_vector)
print("Modified Input:", new_input_vector)

print("Original Output:", output)
print("Modified Output:", input_output)
print("Difference:", input_output - output)
```
