### Demonstrates the logic_gate.py functionality
logic_gate.py uses NumPy to create the logic gates of AND, OR, NAND, NOR, and XOR. It does this by using weights and thresholds to produce an output that equates to the logic gate.

```python
import module3
```

### AND Gate
Uses NumPy dot product with weights and a threshold to Returns 1 when both inputs are 1, and returns 0 when otherwise.

```python
module3.and_gate_table()
```

### OR Gate
Uses NumPy dot product with weights and a threshold to Returns 1 when both or either input is 1, and returns 0 otherwise.

```python
module3.or_gate_table()
```

### NAND Gate
Uses NumPy dot product with weights and a threshold to Returns 1 when BOTH inputs are NOT 1, and returns 0 when otherwise.

```python
module3.nand_gate_table()
```

### NOR Gate
Uses NumPy dot product with weights and a threshold to Returns 1 when NEITHER input is 1, and 0 when otherwise.

```python
module3.nor_gate_table()
```

### XOR Gate
Uses NumPy dot product with weights and a threshold to Returns 1 when both inputs are NOT the same, and 0 when otherwise.

For example, (1 and 1) and (0 and 0) should return 0

```python
module3.xor_gate_table()
```
